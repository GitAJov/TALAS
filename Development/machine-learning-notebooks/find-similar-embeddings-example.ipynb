```python
df = df_a
```

```python

embeddings = np.array(df['embedding'].to_list(), dtype=np.float32)

similarity_threshold = 0.95

similarity_matrix = cosine_similarity(embeddings)

G = nx.Graph()
for i in range(len(similarity_matrix)):
    for j in range(i + 1, len(similarity_matrix)):  # Upper triangular part
        if similarity_matrix[i, j] >= similarity_threshold:
            G.add_edge(i, j)

clusters = [list(component) for component in nx.connected_components(G)]

for i, cluster_indices in enumerate(clusters):
    cluster_titles = df.iloc[cluster_indices]['title'].tolist()
    print(f"Cluster {i+1}:")
    print(cluster_titles)
    print()
```

```python
filtered_clusters = [cluster for cluster in clusters if len(cluster) > 3]

filtered_clusters
```

```python
df['cluster_index'] = -1  # Default to -1 for rows not in any cluster

for cluster_index, cluster_indices in enumerate(clusters):
    df.loc[cluster_indices, 'cluster_index'] = cluster_index
```

```python
import json
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="root",
    database="news"
)

mycursor = mydb.cursor()

# Read the JSON data
with open('../data/sameExample2.json', 'r') as f:
    data = json.load(f)

for item in data:
    # Extract fields from the JSON item
    article_id = item['id']  # Assuming 'id' is included in the JSON
    title = item['title']
    content = item['content']
    source = item['source']
    url = item['url']
    image = "" if item['image'] is None else item['image']
    date = item['date']
    embedding_str = json.dumps(item['embedding'])

    # Prepare the SQL INSERT statement
    sql = """
    INSERT INTO articles (id, title, content, embedding, source, url, image, date) 
    VALUES (%s, %s, %s, %s, %s, %s, %s, %s)
    """
    val = (article_id, title, content, embedding_str, source, url, image, date)

    # Execute the SQL statement
    mycursor.execute(sql, val)

# Commit the changes to the database
mydb.commit()

print(f"Inserted {mycursor.rowcount} rows successfully.")

# Close the cursor and database connection
mycursor.close()
mydb.close()
```
